## FEATURE ENGINEERING 

```python
# Livability Score: higher rooms & better social quality,
# lower crime & poverty = better livability
dataset['LivabilityScore'] = (
    dataset['RM'] * 2
    - dataset['CRIM']
    - dataset['LSTAT']
)

# Location Score: accessibility & pollution impact
dataset['LocationScore'] = (
    dataset['DIS']
    + dataset['RAD']
    - dataset['NOX'] * 10
)

# Property Condition Score
dataset['PropertyScore'] = (
    dataset['RM']
    - dataset['AGE'] * 0.01
    - dataset['TAX'] * 0.001
)

print("Feature engineering complete.")
```

```python
dataset.drop(['B'], axis=1, inplace=True)
```

## Independent and Dependent features

```python
X=dataset.iloc[:,:-1]
y=dataset.iloc[:,-1]
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

## Standardize the dataset

```python
from sklearn.preprocessing import StandardScaler
import pickle

scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

## Feature Column Order

## Model Training

```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
regression.fit(X_train,y_train)
```

## print the coefficients and the intercept

```python
print(regression.coef_)
```

```python
print(regression.intercept_)
```

## on which parameters the model has been trained

```python
regression.get_params()
```

## Prediction With Test Data

```python
reg_pred=regression.predict(X_test)
reg_pred
```

## Assumptions

## plot a scatter plot for the prediction

```python
plt.scatter(y_test,reg_pred)
plt.show()
```

## Residuals

```python
residuals=y_test-reg_pred
residuals
```

## Plot this residuals 

```python
sns.displot(residuals,kind="kde")
plt.show()
```

## Scatter plot with respect to prediction and residuals
## uniform distribution

```python
plt.scatter(reg_pred,residuals)
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error

print(mean_absolute_error(y_test,reg_pred))
print(mean_squared_error(y_test,reg_pred))
print(np.sqrt(mean_squared_error(y_test,reg_pred)))
```

## R square and adjusted R square

# Formula

**R^2 = 1 - SSR/SST**

R^2	=	coefficient of determination
SSR	=	sum of squares of residuals
SST	=	total sum of squares


```python
from sklearn.metrics import r2_score
score=r2_score(y_test,reg_pred)
print(score)
```

## Display adjusted R-squared


**Adjusted R2 = 1 – [(1-R2)*(n-1)/(n-k-1)]**

where:

R2: The R2 of the model
n: The number of observations
k: The number of predictor variables

```python
1 - (1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

## New Data Prediction

```python
sample = X_test[0].reshape(1, -1)
prediction = regression.predict(sample)
print("Predicted Price:", prediction[0])
```

## Pickling The Model file For Deployment

```python
import pickle
pickle.dump(regression,open('regmodel.pkl','wb'))
pickled_model=pickle.load(open('regmodel.pkl','rb'))
```

## Prediction

```python
sample = dataset.drop('Price', axis=1).iloc[0].values.reshape(1, -1)
scaled_sample = scaler.transform(sample)
prediction = pickled_model.predict(scaled_sample)
print("Predicted Price:", prediction[0])
```

## Price Validity Insight

```python
avg_price = dataset['Price'].mean()

if prediction[0] > avg_price * 1.5:
    print("⚠ Price seems overpriced.")
elif prediction[0] < avg_price * 0.5:
    print("⚠ Price seems undervalued.")
else:
    print("✅ Price looks reasonable.")
```
